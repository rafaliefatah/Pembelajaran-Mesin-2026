sinyal, aljabar linier, optimasi, integrasi numerik, statistik, dan manipulasi data sains

3.  Library wandb = Platform MLOps (Machine Learning Operations) yang digunakan untuk melacak eksperimen model pembelajaran
mesin, memvisualisasikan hyperparameter, mencatat metrik performa (accuracy/loss), serta mempermudah kolaborasi antar
peneliti/pengembang

4.  Library pyECG = Library Python khusus yang digunakan untuk analisis dan pemrosesan data sinyal EKG/ECG
(elektrokardiogram). Library ini berfungsi untuk mendeteksi gelombang jantung (seperti gelombang P, QRS complex,
T), ekstraksi fitur sinyal jantung, serta analisis variabilitas detak jantung (Heart Rate Variability / HRV)
